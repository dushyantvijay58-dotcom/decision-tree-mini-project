```python
import pandas as pd
import numpy as numpy
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/loan_approval.csv")
```

```python


x=df[[ 'income', 'credit_score','loan_amount','years_employed']]
y=df['loan_approved']
```

```python
model=DecisionTreeClassifier()
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
acc=accuracy_score(y_test,y_pred)
acc
```

```python
amount = int(input("Enter income: "))
score = int(input("Enter credit score: "))
loan = int(input("Enter loan amount: "))
years = int(input("Enter years employed: "))

pred = model.predict([[amount, score, loan, years]])[0]
print('predicted result: ', pred)
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

features=['income', 'credit_score','loan_amount','years_employed']
plt.figure(figsize=(15,10))
plot_tree(model,feature_names=features,max_depth=3,filled=True,rounded=True,fontsize=12)
plt.show()
```
